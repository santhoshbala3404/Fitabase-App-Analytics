```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dailyact=pd.read_csv("dailyActivity_merged.csv")
dailyact.head(5)
```

```python
dailyact.info()
```

```python
dailyact.describe()
```

```python
plt.hist(dailyact['TotalDistance'])
```

```python
plt.hist(dailyact['Calories'])
```

```python
fig1=plt.figure(figsize=(10,4))
ax1=fig1.add_axes([0,0,1,1])
ax1.plot(dailyact['Id'],dailyact['SedentaryActiveDistance'],label='sedentary active distance')
ax1.plot(dailyact['Id'],dailyact['ModeratelyActiveDistance'],label='moderately active distance')
ax1.plot(dailyact['Id'],dailyact['LightActiveDistance'],label='light active distance')
ax1.plot(dailyact['Id'],dailyact['VeryActiveMinutes'],label='very active distance')
ax1.set_xlabel("ID")
ax1.set_ylabel("Metres")
ax1.legend(loc='best')
```

```python
fig1=plt.figure(figsize=(10,4))
ax1=fig1.add_axes([0,0,1,1])
ax1.plot(dailyact['Id'],dailyact['VeryActiveMinutes'],label='very active minutes')
ax1.plot(dailyact['Id'],dailyact['FairlyActiveMinutes'],label='fairly active minutes')
ax1.plot(dailyact['Id'],dailyact['LightlyActiveMinutes'],label='lightly active minutes')
ax1.plot(dailyact['Id'],dailyact['SedentaryMinutes'],label='sedentary minutes')
ax1.set_xlabel('ID')
ax1.set_ylabel('Minutes')
ax1.legend(loc='best')
```
